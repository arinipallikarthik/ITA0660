```python
import random

TARGET = "11111111"

population_size = 10
chromosome_length = len(TARGET)

# Generate population
population = []

for _ in range(population_size):
    chromosome = ''.join(random.choice('01')
                         for _ in range(chromosome_length))
    population.append(chromosome)

def fitness(chromosome):
    score = 0
    for a,b in zip(chromosome,TARGET):
        if a == b:
            score += 1
    return score

for generation in range(100):

    population = sorted(population,
                        key=fitness,
                        reverse=True)

    if fitness(population[0]) == chromosome_length:
        break

    new_population = population[:2]   # Elitism

    while len(new_population) < population_size:

        parent1 = random.choice(population[:5])
        parent2 = random.choice(population[:5])

        point = random.randint(1, chromosome_length-1)

        child = parent1[:point] + parent2[point:]

        # Mutation
        if random.random() < 0.1:
            pos = random.randint(0, chromosome_length-1)
            child = list(child)
            child[pos] = '1' if child[pos]=='0' else '0'
            child = ''.join(child)

        new_population.append(child)

    population = new_population

print("Best Solution:", population[0])
print("Fitness:", fitness(population[0]))
```
